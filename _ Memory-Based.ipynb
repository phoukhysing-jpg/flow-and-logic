```python
The Virtual Barista (Simple Point-of-Sale)
The Goal: Display a text menu of three coffee shop drinks with different prices. Create a continuous loop that 
asks the user what they want to order. Add the price of their chosen drink to a running total.
If they type an invalid drink, tell them it's not on the menu. The loop should continue until the user 
types "checkout", at which point the program prints their final total.
Concepts Tested: Infinite while loops, break conditions, accumulating totals (+= operator),
complex if/elif/else logic, and string normalization (using .lower() to handle if the user types "Checkout" or 
"CHECKOUT").
```

```python


total = 0

print("------ Welcome to Virtual Barista ------")
print("Menu:")
print("Latte - $3")
print("Espresso - $2")
print("Cappuccino - $4")
print("Type 'checkout' to finish your order.")
print("----------------------------------------")

while True:
    
    drink = input("What would you like to order? ").strip().lower()

    if drink == "checkout":
        break

    if drink == "latte":
        total += 3
        print("Latte added to your order.")

    elif drink == "espresso":
        total += 2
        print("Espresso added to your order.")

    elif drink == "cappuccino":
        total += 4
        print("Cappuccino added to your order.")

    else:
        print("Sorry, that drink is not on the menu.")

print("----------------------------------------")
print("Your final total is: $", total)
print("Thank you for visiting Virtual Barista!")
```
