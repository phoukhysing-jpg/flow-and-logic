Number Guessing Game:
The Goal: The program randomly selects a number between 1 and 100. The user guesses the number,
and the program replies with ""Too high!",  Too low!", or "You got it!". The game loops 
until the user guesses correctly.
Concepts Tested: while loops, if/elif/else statements, comparison operators, importing the basic random module

```python
import random
guess = 1 and 100
secret_number = random.randint(1,100)
print("number beteew 1 and 100")
while guess!=secret_number:
    try:
        guess = int(input("Enter your guess"))
        if guess > secret_number:
            print("Too high")
        elif secret_number > guess:
            print("Too low")
        else:
           print("Your number is correct")
    except ValueError:
        print("Please enter a valid number.")
```
