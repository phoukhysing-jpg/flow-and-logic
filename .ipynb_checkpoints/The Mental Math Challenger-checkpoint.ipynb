The Mental Math Challenger
The Goal: The program generates two random numbers between 1 and 12 and asks the user to multiply them together 
(e.g., "What is 7 x 8?"). If the user gets it right, they get a point, and the program asks another question. 
If they get it wrong,the program tells them the correct answer and ends the game, displaying their final 
streak/score.
Concepts Tested: Importing the random module, tracking score variables, while loops dependent on a 
boolean condition (e.g., while is_correct:), if/else statements, and type casting (converting the user's answerto
an int to compare it to the actual math result).

```python
import random

score = 0
is_correct = True

print("Welcome to the Mental Math Challenger!")

while is_correct:
    num1 = random.randint(1,12)
    num2 = random.randint(1,12)

    answer = int(input(f"What is {num1} x {num2}? "))

    correct_answer = num1 * num2

    if answer == correct_answer:
        print("Correct!")
        score += 1
    else:
        print("Wrong!")
        print("The correct answer was:", correct_answer)
        is_correct = False

print("Game Over!")
print("Your final score:", score)
```
